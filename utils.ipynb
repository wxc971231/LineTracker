```python
import numpy as np
import matplotlib.pyplot as plt

# 可调参数：Huber 损失从二次增长切换为线性增长的阈值
delta = 3.0

# 误差范围，单位可理解为距离 bin（当前也等价于 m）
e = np.linspace(-12, 12, 1000)

# Huber loss
loss = np.where(
    np.abs(e) <= delta,
    0.5 * e**2,
    delta * (np.abs(e) - 0.5 * delta),
)

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.plot(e, loss, lw=2.2, label=fr"Huber loss ($\delta={delta}$)")
ax.axvline(-delta, color="gray", ls="--", lw=1)
ax.axvline(delta, color="gray", ls="--", lw=1, label=r"transition: $|e|=\delta$")
ax.axhline(0, color="black", lw=0.8)

ax.set(
    xlabel=r"error $e=\hat d_\tau-d_\tau$ (bin)",
    ylabel=r"$\mathcal{L}_{\mathrm{Huber}}(e)$",
    title="Huber loss",
)
ax.grid(alpha=0.25)
ax.legend()
plt.show()
```
